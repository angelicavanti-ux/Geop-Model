# Model 3 Geopolitical Specification Comparison

This notebook combines the summary outputs from the restricted macroeconomic baseline and Models 3a–3e into a single comparison dataset.

The specifications compared are:

- Model 3 Restricted: macroeconomic-only baseline
- Model 3a: daily MAX geopolitical indices
- Model 3b: daily SUM geopolitical indices
- Model 3c: 3-day moving averages of the daily SUM indices
- Model 3d: 5-day moving averages of the daily SUM indices
- Model 3e: 10-day moving averages of the daily SUM indices

Each underlying model was estimated over the same 992-observation sample with the final 20 observations reserved for out-of-sample evaluation.

The combined table allows convergence, lag selection, forecasting performance and geopolitical coefficient significance to be compared consistently across specifications.

```python
import pandas as pd

summary_files = [
    "model3a_summary.csv",
    "model3b_summary.csv",
    "model3c_summary.csv",
    "model3d_summary.csv",
    "model3e_summary.csv",
    "model3_restricted_summary.csv"
]

comparison = pd.concat(
    [pd.read_csv(file) for file in summary_files],
    ignore_index=True
)

```

## Comparison Results

The individual model summaries are concatenated into a single table containing convergence diagnostics, information criteria, EUR/USD out-of-sample forecast metrics and geopolitical coefficient p-values.

The comparison shows that Model 3e, using 10-day moving averages of the daily summed geopolitical indices, produces the lowest EUR/USD RMSE and MAE among the specifications considered.

However, none of the geopolitical variables is statistically significant in the EUR/USD equation across Models 3a–3e. The forecasting comparison should therefore be interpreted as evidence about relative model performance rather than evidence that an individual geopolitical coefficient has a statistically significant direct effect on EUR/USD.

```python
comparison.to_csv("model3_geop_comparison.csv", index=False)

print("Saved: model3_geop_comparison.csv")
```

## Summary

This notebook consolidates the restricted Model 3 baseline and the five geopolitical extensions into a single model-comparison dataset.

Among the tested specifications, Model 3e achieves the strongest EUR/USD forecasting performance, with an RMSE of approximately 0.0077 and MAE of approximately 0.0065, compared with approximately 0.0118 and 0.0104 respectively for the restricted macroeconomic baseline.

The comparison therefore indicates that the 10-day smoothed geopolitical specification improves forecast accuracy relative to the comparable macroeconomic-only model over the common holdout period. At the same time, the geopolitical coefficients are not individually statistically significant in the EUR/USD equation, so the results do not provide strong evidence of direct standalone predictive effects from any single geopolitical index.

The consolidated comparison is exported as `model3_geop_comparison.csv` for use in the final results analysis and dissertation write-up.
