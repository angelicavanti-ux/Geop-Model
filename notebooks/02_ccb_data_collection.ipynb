# Cross-Currency Basis Data Availability Assessment

## Purpose

The original project design considered EUR/USD cross-currency basis (CCB) as the primary market variable of interest. However, the available Bloomberg CCB series were found to have insufficient historical coverage for the geopolitical sample constructed from Donald Trump's first-presidency tweets.

This notebook documents the available EUR/USD CCB data and assesses whether the series provide sufficient temporal overlap with the geopolitical dataset.

The available Bloomberg series are:

- EUR/USD 3-month cross-currency basis
- EUR/USD 12-month cross-currency basis

Both series begin in February 2021, whereas the Trump tweet sample ends in January 2021. Consequently, the datasets do not provide a usable overlapping sample for estimation of the proposed geopolitical forecasting model.

EUR/USD spot movements (DEXUSEU) were therefore adopted as the primary modelling target in the subsequent analysis.

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller
```

## 1. Import Available CCB Data

The available EUR/USD cross-currency basis series were extracted from Bloomberg Terminal.

The datasets represent the 3-month and 12-month EUR/USD basis swaps using ESTRON and SOFR-based reference rates. Historical availability for these series begins in February 2021.

```python
ccb3m = pd.read_csv("../data/raw/EURUSD_Basis_Swap_ESTRONvsSOFRRATE_3M.csv", parse_dates=True, index_col=0)

ccb12m = pd.read_csv("../data/raw/EURUSD_ESTRONvsSOFRRATE_12M.csv", parse_dates=True, index_col=0)       

```

## 2. Initial Inspection

```python
print("CCB 3M")
print("Shape:", ccb3m.shape)
print("Date range:", ccb3m.index.min(), "to", ccb3m.index.max())
print(ccb3m.head())

print("\nCCB 12M")
print("Shape:", ccb12m.shape)
print("Date range:", ccb12m.index.min(), "to", ccb12m.index.max())
print(ccb12m.head())
```

## 3. Assessment of Temporal Overlap

The geopolitical dataset used in the main analysis is based on Donald Trump's first-presidency tweets and ends in January 2021.

The available EUR/USD CCB series begin only in February 2021. Therefore, there is no meaningful overlapping estimation period between the available CCB data and the geopolitical tweet sample.

Using CCB as either the primary modelling target or a later backtesting variable would consequently require a different geopolitical dataset and would introduce additional complexity outside the scope of the present analysis.

For this reason, CCB is not used in the subsequent modelling pipeline. EUR/USD spot movements are instead used as the primary target variable.
